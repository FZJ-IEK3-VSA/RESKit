```python
import reskit as rk
```

## Create LRA from ERA5 data directly (e.g. 100m ws or GHI)

```python
# GHI

# weather_data source
base_path = "path_to_ERA5_data"
zoom_level = 4 # only needed if data is tiled.

# file name of relevant dataset
variable = "surface_solar_radiation_downwards.processed.t_adjusted"

# filename of resulting LRA dataset
variable_name_output = "GHI"

# Directories
out_dir = "some_path"
temp_dir = "some_path" # each year is first calculated seperately

# start and end year based of which the LRA is created
start_year = 1994
end_year = 2018

rk.create_LRA(
    base_path=base_path,
    variable=variable,
    variable_name_output=variable_name_output,
    start_year=start_year,
    end_year=end_year,
    zoom_level=zoom_level,
    out_dir=out_dir,
    temp_dir=temp_dir,
    weather_source_prefix="ERA5",
    write_geotiff=True,
    write_netcdf=True,
    cache_yearly=True,
)
```

## Create LRA for DNI (needs additional calculation)

```python
# DNI

# weather_data source
base_path = "path_to_ERA5_data"
zoom_level = 4 # only needed if data is tiled.

# file name of relevant datasets needed for calculation
direct_horizontal_irradiance_variable = "total_sky_direct_solar_radiation_at_surface.processed.t_adjusted"
surface_temperature_variable = "2m_temperature"
surface_pressure_variable = "surface_pressure"

# filename of resulting LRA dataset
variable_name_output = "DNI"
# filename of temporary data output
variable = "direct_normal_irradiance"

# Directories
out_dir = "some_path"
temp_dir = "some_path" # each year is first calculated seperately

# start and end year based of which the LRA is created
start_year = 1994
end_year = 2018

rk.create_DNI_LRA(
    base_path=base_path,
    variable=variable,
    direct_horizontal_irradiance_variable=direct_horizontal_irradiance_variable,
    variable_name_output=variable_name_output,
    surface_temperature_variable=surface_temperature_variable,
    surface_pressure_variable=surface_pressure_variable,
    start_year=start_year,
    end_year=end_year,
    zoom_level=zoom_level,
    out_dir=out_dir,
    temp_dir=temp_dir,
    weather_source_prefix="ERA5",
    write_geotiff=True,
    write_netcdf=True,
    cache_yearly=True,
)
```
